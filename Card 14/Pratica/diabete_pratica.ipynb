```python
import pandas as pd 
import tensorflow as tf 
import sklearn 

from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Dense 
from sklearn.metrics import accuracy_score, confusion_matrix  #imports
```

```python
base = pd.read_csv('diabetes.csv') #le o csv
base
```

```python
x = base.drop('Outcome', axis=1) #atributos previsores
y = base['Outcome'] #classe q tem o resultado binario
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.25) #divide os dados em treino e teste
```

```python
classificador = Sequential() #criando a rede neural
classificador.add(Dense(units = 8, activation = 'relu', kernel_initializer = 'random_uniform', input_dim = 8)) #camada entrada
classificador.add(Dense(units = 1, activation = 'sigmoid')) #camada de saida
```

```python
classificador.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['binary_accuracy']) #compilando a rede
```

```python
classificador.fit(X_train, Y_train, batch_size = 10, epochs = 100) #treinando ela
```

```python
previsoes = classificador.predict(X_test) #fazendo as previsoes
previsoes = previsoes > 0.5 #transformando em true ou false
```

```python
previsoes #as previsoes em si
```

```python
Y_test #valores reais
```

```python
accuracy_score(Y_test, previsoes) #acuracia
```

```python
confusion_matrix(Y_test, previsoes) #matriz de confusao
```
