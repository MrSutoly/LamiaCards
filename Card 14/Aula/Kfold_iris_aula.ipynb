```python
import pandas as pd
import tensorflow as tf
import sklearn
import scikeras
from scikeras.wrappers import KerasClassifier
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras import utils as np_utils
from tensorflow.keras import backend as k
from tensorflow.keras.models import Sequential
from sklearn.model_selection import cross_val_score#imports
```

```python
base = pd.read_csv('iris.csv')
base #lendo csv e jogando pra variavel
```

```python
X = base.iloc[:, 0:4].values #pegando apenas alguns valores
```

```python
y = base.iloc[:, 4].values#pegando apenas alguns valores
```

```python
le = LabelEncoder() #joga pra matrix binaria
y = le.fit_transform(y)
y = np_utils.to_categorical(y) 
```

```python
def criar_rede():
    k.clear_session()
    rede_neural = Sequential([
        tf.keras.layers.InputLayer(shape=(4,)),
        tf.keras.layers.Dense(units=4, activation='relu'),
        tf.keras.layers.Dense(units=4, activation='relu'),
        tf.keras.layers.Dense(units=3, activation='softmax')
    ])
    rede_neural.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['categorical_accuracy'])
    return rede_neural #cria a rede neural
```

```python
rede_neural = KerasClassifier(model = criar_rede, epochs = 250, batch_size = 10) #colocando os parametros
```

```python
resultados = cross_val_score(estimator = rede_neural, X = X, y = y, cv = 10, scoring = 'accuracy') #treinando com KFOLD
```
