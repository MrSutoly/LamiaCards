```python
import pandas as pd #import
```

```python
x = pd.read_csv('entradas_breast.csv') #le o csv
```

```python
x #print do csv
```

```python
y = pd.read_csv('saidas_breast.csv')
y #le e printa
```

```python
import sklearn
from sklearn.model_selection import train_test_split #imports
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size = 0.25) #divide os dados em treino e teste
```

```python
X_train.shape #o tamanho do X de treino
```

```python
X_test.shape #tamanho do X de teste
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential #imports
```

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(shape = (30,)),
    tf.keras.layers.Dense(units = 16, activation = 'relu', kernel_initializer = 'random_uniform'),
    tf.keras.layers.Dense(units = 1, activation = 'sigmoid')
]) #criando a rede neural em si com os algoritimos e a camada oculta
```

```python
rede_neural.summary() #ver como a rede neural esta criada
```

```python
rede_neural.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['binary_accuracy']) #compilando a rede
```

```python
rede_neural.fit(X_train, Y_train, batch_size = 10, epochs = 100) #treinando ela
```

```python
previsoes = rede_neural.predict(X_test) #fazendo as previsoes
```

```python
previsoes = previsoes > 0.5 #pegando so as acima de 0.5
```

```python
previsoes #a linha anterior jogou os valores pra true ou false pra comparar com o valor real do Y q esta embaixo
```

```python
Y_test #Y real pra comparar com o X
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix #imports
```

```python
accuracy_score(Y_test, previsoes) #vendo a acuracia
```

```python
confusion_matrix(Y_test, previsoes) #jogando pra matriz de confusao
```

```python
49 + 81, 3 + 10
```

```python
130 / 143, 13 / 143 #somas e divisoes pra mostrar como foi calculado os valores anteriores
```

```python
rede_neural.evaluate(X_test, Y_test) #vendo quanto ela pontuou comparando os valores
```

# Adicionando mais camadas

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(shape = (30,)),
    tf.keras.layers.Dense(units = 16, activation = 'relu', kernel_initializer = 'random_uniform'),
    tf.keras.layers.Dense(units = 16, activation = 'relu', kernel_initializer = 'random_uniform'),
    tf.keras.layers.Dense(units = 1, activation = 'sigmoid')
]) #criando mais uma camada oculta
```

```python
rede_neural.summary() #msm coisa denovo, vendo como ela foi criada
```

```python
otimizador = tf.keras.optimizers.Adam(learning_rate = 0.001, clipvalue = 0.5) #defiinindo um otimizador
```

```python
rede_neural.compile(optimizer = otimizador, loss = 'binary_crossentropy', metrics = ['binary_accuracy']) #compilando ela usando o otimizador anterior
```

```python
rede_neural.fit(X_train, Y_train, batch_size = 10, epochs = 100) #treinando 
```

```python
rede_neural.evaluate(X_test, Y_test) #pontuacao
```
