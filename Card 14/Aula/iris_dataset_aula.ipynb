```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils as np_utils
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix 
from sklearn.metrics import accuracy_score #imports
```

```python
base = pd.read_csv('iris.csv')
base #lendo csv e jogando pra variavel
```

```python
X = base.iloc[:, 0:4].values #pegando apenas alguns valores
X
```

```python
y = base.iloc[:, 4].values
y #pegando apenas alguns valores
```

```python
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)
y #jogando pra matriz binaria
```

```python
y.shape #tamanho
```

```python
y = np_utils.to_categorical(y) #jogando pro formato de matriz com 3 colunas
y
```

```python
y.shape #tamanho
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25) #splitando os dados
```

```python
X_train.shape, X_test.shape #vendo tamanho de X
```

```python
y_train.shape, y_test.shape #vendo tamanho de Y
```

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(shape = (4,)),
    tf.keras.layers.Dense(units = 4, activation = 'relu'), 
    tf.keras.layers.Dense(units = 4, activation = 'relu'), 
    tf.keras.layers.Dense(units = 3, activation = 'softmax') 
]) #criando a rede com as camadas
```

```python
rede_neural.summary() #vendo como a rede e
```

```python
rede_neural.compile(optimizer = 'adam', loss = 'categorical_crossentropy', metrics = ['categorical_accuracy']) #compilando
```

```python
rede_neural.fit(X_train, y_train, batch_size = 10, epochs = 1000) #treinando
```

```python
rede_neural.evaluate(X_test, y_test) #estatica pros X e Y de teste
```

```python
previsao = rede_neural.predict(X_test) #previsao pra X teste
```

```python
previsao = previsao > 0.5 #jogando pra booleano
previsao
```

```python
y_test2 = [np.argmax(t) for t in y_test] #index das respostas
```

```python
y_test2
```

```python
previsao2 = [np.argmax(t) for t in previsao] #mesma coisa
```
