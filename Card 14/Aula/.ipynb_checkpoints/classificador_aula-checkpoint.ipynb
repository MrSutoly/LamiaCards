```python
import pandas as pd
import numpy as np
import tensorflow as tf #imports
```

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
import tensorflow as tf
print("Dispositivos encontrados:", tf.config.list_physical_devices()) #tive q fazer isso pois o fit nao estava funcionando
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, InputLayer, Dropout #imports
```

```python
X = pd.read_csv('entradas_breast.csv')
y=  pd.read_csv('saidas_breast.csv') #dados
```

```python
classificador = Sequential()
classificador.add(InputLayer(shape=(30,)))
classificador.add(Dense(units=16, activation='relu', kernel_initializer='random_uniform'))
classificador.add(Dropout(0.2))
classificador.add(Dense(units=16, activation='relu', kernel_initializer='random_uniform'))
classificador.add(Dropout(0.2))
classificador.add(Dense(units=1, activation='sigmoid')) #classificador e as camadas dele
```

```python
classificador.compile(optimizer='adam', loss='binary_crossentropy', metrics=['binary_accuracy']) #compila
```

```python
classificador.fit(X, y, batch_size = 10, epochs = 100) #treinando
```

```python
novo = np.array([[15.80, 8.34, 118, 900, 0.10, 0.26, 0.08, 0.134, 0.178, 0.20, 0.05, 1098, 0.87, 4500, 145.2, 0.005, 0.04, 0.05, 0.015, 0.03,
0.007, 23.15, 16.64, 178.5, 2018, 0.14, 0.185, 0.84, 158, 0.363]]) #novo dado
```

```python
novo.size #tamanho do novo dado
```

```python
previsao = classificador.predict(novo) #fazendo a previsao pro novo dado
```

```python
previsao = previsao > 0.5
previsao #fazendo ficar True ou False
```

```python
if previsao: #pra ver se e maligno ou nao
    print('Malígno')
else: 
    print('Benigno')
```
