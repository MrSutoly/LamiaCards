Step Function

```python
def step_function(soma): #funcao se o numero for maior q 1 ele retorna 1 se menor ele volta 0
  if soma >= 1:
    return 1
  return 0
```

```python
step_function(-10) #testando
```

```python
step_function(10) #testando
```

Sigmoid

```python
def sigmoid_function(soma): #funcao q pega o exponencial da soma negativa pra colocar entre 0 e 1
  return 1 / (1 + np.exp(-soma))
```

```python
sigmoid_function(-10) #testando
```

```python
sigmoid_function(10) #testando
```

```python
sigmoid_function(1) #testando
```

Hyperbolic

```python
def tahn_function(soma):
  return (np.exp(soma) - np.exp(-soma)) / (np.exp(soma) + np.exp(-soma)) #pega qualquer valor e joga entre -1 e 1
```

```python
tahn_function(-10) #testando
```

```python
tahn_function(10) #testando
```

```python
tahn_function(1) #testando
```
