```python
valores = {'vehicleType': 'limousine',
    'gearbox': 'manuell',
    'model': 'golf',
    'fuelType': 'benzin',
    'notRepairedDamage': 'nein'
} #valores
```

```python
base = base.fillna(value = valores) #substituindo nos NaN
```

```python
base
```

```python
base.isnull().sum() #nenhum valor null
```

```python
X = base.iloc[:, 1:12].values
X #pegando as colunas e linhas
```

```python
y = base.iloc[:, 0].values
y #vendo todas as linhas da coluna 0
```

```python
base['brand'].value_counts() #vendo as marcas
```

```python
onehotencoder = ColumnTransformer(transformers=[('OneHot', OneHotEncoder(), [0, 1, 3, 5, 8, 9, 10])], remainder='passthrough') #passando dado categorico pra numerico
```

```python
X = onehotencoder.fit_transform(X).toarray() #transforma pra array
X
```

```python
base.shape #base original
```

```python
X.shape #novas colunas
```

```python
regressor = Sequential([
    tf.keras.layers.InputLayer(shape = (316,)),
    tf.keras.layers.Dense(units = 158, activation = 'relu'),
    tf.keras.layers.Dense(units = 158, activation = 'relu'),
    tf.keras.layers.Dense(units = 1, activation = 'linear')
]) #criando a rede de regressao
```

```python
regressor.summary() #sumario da rede
```

```python
regressor.compile(loss='mean_absolute_error', optimizer='adam', metrics=['mean_absolute_error']) #compilando
```

```python
regressor.fit(X, y, batch_size = 300, epochs = 100) #treinando
```

```python
previsao = regressor.predict(X) #prevendo pra X
```

```python
previsao #vendo os valores
```
