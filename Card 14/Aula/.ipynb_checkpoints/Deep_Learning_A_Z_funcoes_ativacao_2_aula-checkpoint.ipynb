ReLU

```python
def relu_function(soma): #se for maior q 0 ele volta o proprio numero
  if soma >= 0:
    return soma
  return 0
```

```python
relu_function(-20) #testando
```

```python
relu_function(-10) #testando
```

```python
relu_function(10) #testando
```

Softmax

```python
import numpy as np
def softmax_function(x):
  ex = np.exp(x)
  return ex / ex.sum() #joga pra exponencial e divide pela soma dos nmrs numa lista
```

```python
softmax_function([7.0, 2.0, 1.3]) #pra cada valor ele da uma probabilidade
```
