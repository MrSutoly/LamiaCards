```python
import re
#formata as log do site do video
format_pat = re.compile(
    r'(?P<host>[\d\.]+)\s'
    r'(?P<identity>\S*)\s'
    r'(?P<user>\S*)\s'
    r'\[(?P<time>.*?)\]\s'
    r'"(?P<request>.*?)"\s'
    r'(?P<status>\d+)\s'
    r'(?P<bytes>\S*)\s'
    r'"(?P<referer>.*?)"\s'
    r'"(?P<user_agent>.*?)"\s*'
)
```

```python
logPath = "../anexos/Aquivos_de_C%C3%B3digo/access_log.txt" #caminho pro arquivo com as logs
```

```python
URLCounts = {}
feed_logs = [] 

with open(logPath, "r") as f:
    for line in (l.rstrip() for l in f):
        match = format_pat.match(line)
        if match:
            access = match.groupdict() 
            agent = access['user_agent'] #acessa user_agent
            request = access['request'] #acessa request

            fields = request.split() #quebra a request

            if len(fields) == 3: #se for 3
                (action, URL, protocol) = fields #joga pras variaveis
                
                if "feed" in URL: #se tiver feed joga pra pasta com a URL e o browser
                    feed_logs.append((URL, agent))

                if (not ('bot' in agent or 'spider' in agent or
                         'Bot' in agent or 'W3 Total Cache' in agent or agent == '-')): #se nao e bot

                    if (URL.endswith("/") and "feed" not in URL): #se acabar com contra barra e n tiver feed na URL
                        if action == 'GET': #e se for
                            if URL in URLCounts: #soma +1 ou add ela no dicionario
                                URLCounts[URL] += 1
                            else:
                                URLCounts[URL] = 1

results = sorted(URLCounts, key = lambda i: int(URLCounts[i]), reverse = True) #joga do maior pro menor

print("URL sem feed")
for result in results[:20]:
    print(result + ": " + str(URLCounts[result]))

print("")

print("Analise dos acessos:")
for URL, agent in feed_logs[:10]:  #mostra so os 10 primeiros
    print("URL:", URL)
    print("User-Agent:", agent)
```
