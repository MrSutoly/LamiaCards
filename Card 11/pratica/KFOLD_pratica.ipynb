```python
import numpy as np
#horas de estudo e faltas
X = np.array([
    [1, 10],
    [2, 8],
    [3, 5],
    [4, 4],
    [5, 2],
    [6, 1],
    [2, 9],
    [3, 6]
])

#0 é reprovado e 1 é aprovado
y = np.array([0, 0, 0, 1, 1, 1, 0, 0])
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.neighbors import KNeighborsClassifier
#import

kf = KFold(n_splits = 4, shuffle = True, random_state = 42) #divide em 4 pastas 

modelo = KNeighborsClassifier(n_neighbors = 3)  #3 vizinhos

for i, (train_index, test_index) in enumerate(kf.split(X)): #vai passando por cada pasta pegando os dados de cada uma
    print(f"{i+1} Pasta:") #qual pasta ele ta
    
    X_train, X_test = X[train_index], X[test_index] #pra cada pasta o X_train e X_test e Y_train e Y_test mudam
    y_train, y_test = y[train_index], y[test_index]
    
    modelo.fit(X_train, y_train) #treina com os dados
    y_pred = modelo.predict(X_test) #preve pra aquela pasta

    #prints
    print("Dados de teste", X_test)
    print("Dado Real", y_test)
    print("Previsto pelo modelo:", y_pred)

    acc = np.mean(y_pred == y_test) #acuracia (o tanto q acertou)
    print("Acuracia:", acc)
    print('')
```
