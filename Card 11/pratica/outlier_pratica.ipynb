Intead of a single outlier, add a several randomly-generated outliers to the data. Experiment with different values of the multiple of the standard deviation to identify outliers, adn see what effect it has on the final results

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
```

```python
incomes = np.random.normal(24000, 70000, 12000) #media 24000, dp de 70000 e 12000 dados
outliers = np.random.normal(1000000, 500000, 20)  #gera 20 outliers
incomes = np.append(incomes, outliers) #joga os dados pro incomes
plt.hist(incomes, 50)
plt.show()
```

```python
u = np.mean(incomes) #media
s = np.std(incomes) #desvio padrao
```

```python
filtered = [e for e in incomes if (u - 1*s < e < u + 1*s)] #desvio de sino so pega os dados  estao a 1 Dp de distancia
```

```python
plt.hist(filtered, 50) #histograma
plt.show()
```

```python
filtered = [e for e in incomes if (u - 3*s < e < u + 3*s)] #desvio de sino so pega os dados  estao a 3 Dp de distancia
plt.hist(filtered, 50) #histograma
plt.show()
```
