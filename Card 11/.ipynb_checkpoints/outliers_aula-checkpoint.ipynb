```python
%matplotlib inline
import numpy as np

incomes = np.random.normal(27000, 15000, 10000) #media 2700, dp de 15000 e 10000 dados
incomes = np.append(incomes, [1000000000]) #add um bilionario na lista

import matplotlib.pyplot as plt
plt.hist(incomes, 50)
plt.show()
```

```python
incomes.mean() #media da lista
```

```python
def reject_outliers(data):
    u = np.median(data) #pega a mediana dos dados
    s = np.std(data) #desvio padrao dos dados
    filtered = [e for e in data if (u - 2 * s <e < u + 2 * s)] #define o limite minimo e maximo
    return filtered

filtered = reject_outliers(incomes) #chama a funcao

plt.hist(filtered, 50) #histograma
plt.show()
```

```python
np.mean(filtered) #media dos dados filtrados
```
