```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
import pylab as pl
from itertools import cycle

iris = load_iris() #carrega as flores

numSamples, numFeatures = iris.data.shape #pega linhas e colunas
print(numSamples)
print(numFeatures)
print (list(iris.target_names)) #pega os nomes
```

```python
X = iris.data
pca = PCA(n_components = 2, whiten = True).fit(X) #reduzindo de 4 pra 2 dimensoes e o whiten normaliza os dados
X_pca = pca.transform(X) #aplica a transformacao
```

```python
print(pca.components_) #os pesos pra transformar de 4 pra 2
```

```python
print(pca.explained_variance_ratio_) #quanto de informacao ficou nas novas dimensoes
print(sum(pca.explained_variance_ratio_)) #quanto da informacao original ficou
```

```python
colors = cycle('rgb')
target_ids = range(len(iris.target_names)) #pega o nome das especies
pl.figure()
for i, c, label in zip(target_ids, colors, iris.target_names): #define cores pra cada especie
    pl.scatter(X_pca[iris.target == i, 0], X_pca[iris.target == i, 1], #pega os dados de uma espcie de cada vez e gera o grafico
             c = c, label = label)
pl.legend()
pl.show()
```
