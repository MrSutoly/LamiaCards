The poly kernel for SVC actually has another attribute for the number of degress of the polynomial used, wich defaults to 3. For example svm.SVC(kernel = 'poly', degree = 3, C = 1)

```python
import numpy as np
from sklearn import model_selection
from sklearn import datasets
from sklearn import svm

iris = datasets.load_iris() #carrega o dataset com fotos de flor
```

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(iris.data, iris.target, test_size = 0.4, random_state = 0) #cada variavel com cada tipo de dado

clf = svm.SVC(kernel = 'linear', C = 1).fit(X_train, y_train) #treinando o modelo

clf.score(X_test, y_test) #testando o modelo
```

```python
clf = svm.SVC(kernel = 'poly', C = 1).fit(X_train, y_train) #treina o modelo degree em 3
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv = 5) #faz o kfold
print(scores)
print(scores.mean())
```

```python
clf = svm.SVC(kernel = 'poly', C = 1, degree = 6).fit(X_train, y_train) #treina o modelo com degree em 6
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv = 5) #faz o kfold
print(scores)
print(scores.mean())
```

```python
clf = svm.SVC(kernel = 'poly', C = 1, degree = 8).fit(X_train, y_train) #treina o modelo com degree em 8
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv = 5) #faz o kfold
print(scores)
print(scores.mean())
```
