```python
from sklearn.datasets import load_iris  #importa o dataset de flor iris

iris = load_iris() #carrega o dataset

numSamples, numFeatures = iris.data.shape #pega as colunas do dataset
print(numSamples)
print(numFeatures)
print(list(iris.target_names)) #printa o nome e o nmr de cada um com o print de cima
```

```python
from sklearn.model_selection import train_test_split #import

X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size = 0.2, random_state = 0) #separa em dados de treino e teste
```

```python
import xgboost as xgb #import do xgb

train = xgb.DMatrix(X_train, label = y_train) #pega o dado de treino e joga pro dmatrix do import de cima
test = xgb.DMatrix(X_test, label = y_test) #mesma coisa agr com os dados de teste
```

```python
param = {
    'max_depth' : 4,
    'eta' : 0.3,
    'objective' : 'multi:softmax',
    'num_class' : 3} #dicionario com as cfgs pra treinar o modelo
epochs = 10 #qnts vezes ele vai repetir pra aprender
```

```python
model = xgb.train(param, train, epochs) #treina com tudo de antes
```

```python
predictions = model.predict(test) #testa ele com os dados de test
```

```python
print(predictions) #printa os resultados
```

```python
from sklearn.metrics import accuracy_score #import

accuracy_score(y_test, predictions) #fez a acuracia do teste
```
