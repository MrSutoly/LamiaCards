```python
import numpy as np
import pandas as pd
from sklearn import tree #imports

input_file = "../emails/PastHires.csv" #le o arquivo dentro da pasta
df = pd.read_csv(input_file, header = 0) #le o csv e joga pra um DF
```

```python
df.head()
```

```python
d = {'Y': 1, 'N': 0} #dicionario pra Y ser 1 e N ser 0
df['Hired'] = df['Hired'].map(d) #Cjoga a coluna hired que tem o Y e N pra 1 e 0
df['Employed?'] = df['Employed?'].map(d) #msm coisa pra essa
df['Top-tier school'] = df['Top-tier school'].map(d) #msm coisa pra essa
df['Interned'] = df['Interned'].map(d) #msm coisa com outra coluna 
d = {'BS': 0, 'MS': 1, 'PhD': 2} #se tem bs 0 ms 1 e phd 2 igual com o Y e N antes 
df['Level of Education'] = df['Level of Education'].map(d) #msm coisa de antes agr pros 3 de antes
df.head()
```

```python
features = list(df.columns[:6])  #joga as 6 primeiras colunas do DF pra lista
features #printa
```

```python
y = df["Hired"] #define sque vamos prever se a pessoa foi contratada ou nao
X = df[features] #entradas do modelo 
clf = tree.DecisionTreeClassifier() #faz o modelo da arvore
clf = clf.fit(X, y) #treina
```

```python
from IPython.display import Image
from io import StringIO
import pydot #imports

dot_data = StringIO() #armezenar o grafico com esse buffer de texto
tree.export_graphviz(clf, out_file = dot_data, feature_names = features) #joga pro formato visual

graph = pydot.graph_from_dot_data(dot_data.getvalue()) #pega o texto e joga pra grafico
Image(graph[0].create_png()) #cria a imagem
```

```python
from sklearn.ensemble import RandomForestClassifier #import

clf = RandomForestClassifier(n_estimators = 10) #cria 10 arvores de decisao
clf = clf.fit(X, y) #treina 

print(clf.predict([[10, 1, 4, 0, 0, 0]]))
print(clf.predict([[10, 0, 4, 0, 0, 0]])) #prints
```
