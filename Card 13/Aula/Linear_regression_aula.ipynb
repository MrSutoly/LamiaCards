```python
import numpy as np
import matplotlib.pyplot as plt #imports 
```

```python
pageSpeeds = np.random.normal(3.0, 1.0, 1000) #faz uma distribuicao normal com media 3 dp de 1
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 0.1, 1000)) * 3 #valores de compra 

plt.scatter(pageSpeeds, purchaseAmount) #faz o grafico
plt.show()
```

```python
from scipy import stats #importa scipy 

slope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount) #pega os stats, tipo, o slope e o quanto o y muda a partir de x aumenta, o intercept e valor de y quando x e zero, pega o valor de R.....
```

```python
r_value ** 2 #valor R ao quadrado
```

```python
def predict(x): #funcao pra 
    return slope * x + intercept #faz o slope * x + o intercept

fitLine = predict(pageSpeeds) #usa a funcao no vetor de velocidade de carregamento das paginas

plt.scatter(pageSpeeds, purchaseAmount) #faz o grafico
plt.plot(pageSpeeds, fitLine, c ='r') #faz a linha vermelha
plt.show()
```
