```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences #imports
```

```python
sentences = [
    "I like eggs and ham.",
    "I love chocolate and bunnies.",
    "I hate onions."
] #palavras
```

```python
MAX_VOCAB_SIZE = 20000 #tamanho das palavras
tokenizer = Tokenizer(num_words = MAX_VOCAB_SIZE) #define o tokenizer e a qntd de pelavras nele
tokenizer.fit_on_texts(sentences) #colocando as palavras
sequences = tokenizer.texts_to_sequences(sentences) #jogando pra inteiro
```

```python
sequences
```

```python
tokenizer.word_index #pegando cada int de cada palavra
```

```python
data = pad_sequences(sequences) #padding pra preencher e deixar tudo do msm tamanho
data
```

```python
MAX_SEQUENCE_LENGTH = 5
data = pad_sequences(sequences, maxlen = MAX_SEQUENCE_LENGTH) #tamanho maximo de 5
data
```

```python
data = pad_sequences(sequences, maxlen = MAX_SEQUENCE_LENGTH, padding = 'post') #agora os 0 sao no final
data
```

```python
data = pad_sequences(sequences, maxlen = 6) #tamanho maximo de 6
data #onde n tem nada ele coloca o 0 pra fechar o tamanho de 6
```

```python
data = pad_sequences(sequences, maxlen = 4) #cortando pra fechar 4
data
```

```python
data = pad_sequences(sequences, maxlen = 4, truncating = 'post') #agora ele corta no final pra fechar 4
data
```
