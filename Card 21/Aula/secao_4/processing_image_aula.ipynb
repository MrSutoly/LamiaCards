```python
img = cv.imread('ab6761610000e5eb31f6ab67e6025de876475814')
```

```python
rgb_image = cv.cvtColor(img, cv.COLOR_BGR2RGB) #deixa a iamgem original
plt.imshow(rgb_image)
```

```python
gray_image = cv.cvtColor(rgb_image, cv.COLOR_RGB2GRAY)
plt.imshow(gray_image, cmap='gray') #deixa a imagem em cinza
```
