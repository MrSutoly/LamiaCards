```python
import math #import

def sigmoid(x):
    return 1 / (1 + math.exp(-x)) #funcao sigmoide onde receve x e divide 1 pelo expoente negativo do x
```

```python
import torch #import

class ReLU(torch.nn.Module):  #funcao relu
    def __init__(self):
        super().__init__()
    
    def forward(self, x):
        return torch.max(torch.zeros_like(x), x) #compara a matriz de 0 feita e se na matriz original tinha um numero nao negativo ele e colocado na matriz q vai ser retornada
```

```python
import numpy as np #import

def cross_entropy_loss(y_true, y_pred): #calcula o erro do valor previsto e do real
    epsilon = 1e-15 #epsilon pra n ter log de 0
    y_pred = np.clip(y_pred, epsilon, 1. - epsilon) #
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)) #calcula a media da cross entropy
```

```python
def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2) #eleva ao quadrado pra eliminar valor negativo
```
