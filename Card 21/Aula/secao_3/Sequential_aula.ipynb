```python
from torch.utils.data import Dataset, DataLoader
import torch
import numpy as np
import torch.nn as nn 
import time
from torch.optim import SGD
#imports
```

```python
x = [[1, 2], [3, 4], [5, 6], [7, 8]] #dados pro treino, onde y e a soma dos numeros nos pares
y = [[3], [7], [11], [15]]
```

```python
class MyDataset(Dataset): #cria o dataset
    def __init__(self, x, y):
        self.x = torch.tensor(x).float() 
        self.y = torch.tensor(y).float() #converte pra tensor e .float
    def __len__(self):
        return len(self.x) #tamanho do dataset
    def __getitem__(self, ix): #pega um nmr no x e no y
        return self.x[ix], self.y[ix]
```

```python
ds = MyDataset(x, y) #coloca o dataset na variavel ds
dl = DataLoader(ds, batch_size = 2, shuffle = True) #define q tem 2 grupo de dado e embaralha
```

```python
model = nn.Sequential(nn.Linear(2,8),
                      nn.ReLU(),
                      nn.Linear(8,1)) #modelo sequencial com as camadas
```

```python
model #camada linear com 2 input e 8 output a de atiavacao e a outra q pega as 8 saidas e joga pra 1 output so
```

```python
loss_func = nn.MSELoss() #funcao de perda
opt = SGD(model.parameters(), lr= 0.001) #optimizador

start = time.time() #inicia

model.train() #treina o modelo

loss_history = [] #lista pra perda
for _ in range(50): #50 epochs
    for ix, iy in dl: 
        opt.zero_grad() #zera os gradiente pra ir pra nova epoch
        loss_value = loss_func(model(ix), iy) #valor do erro 
        loss_value.backward()
        opt.step() #atualiza
        loss_history.append(loss_value)
end = time.time() #acaba o timer
print(end - start)
```
