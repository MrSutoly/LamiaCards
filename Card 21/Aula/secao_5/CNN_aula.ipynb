```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'

#dados de treino
X_train = torch.tensor([[[[4, 5, 6, 7], [5, 6, 7, 8], [8, 9, 10, 11], [4, 6, 7, 8]]], [[[-4, 5, 6, -7], [5, -6, 7, 8], [-8, 9, -10, 11], [-4, -6, -7, -8]]]]).float().to(device)
X_train.div_(8) #divide o tensor
y_train = torch.tensor([0, 1]).float().to(device) #dado de treino y
```

```python
X_train.shape #formato do x
```

```python
y_train.shape #formato do y
```

```python
import torch
import torch.nn as nn
from torch.optim import Adam #imports

def get_model():
    model = nn.Sequential( #modelo
        nn.Conv2d(in_channels = 1, out_channels = 1, kernel_size = 3), #camada de convolucao
        nn.MaxPool2d(kernel_size = 2), #faz o maxpooling
        nn.ReLU(), #ativacao
        nn.Flatten(), #passa pra vetor
        nn.Linear(1,1), #conecta
        nn.Sigmoid() #atuvacao 0 ou 1
    ).to(device)
    loss_fn = nn.BCELoss() #perda
    optimizer = Adam(model.parameters(), lr = 0.01) #otimizador
    return model, loss_fn, optimizer
```

```python
model, criterion, optimizer = get_model() #chama o modelo
```

```python
model #print
```

```python
import torch #import

def train_batch(x, y, model, optimizer, loss_fn):
    model.train() #treina
    optimizer.zero_grad() #zera o gradiente antes de ir pro proximo
    prediction = model(x) #preve
    batch_loss = loss_fn(prediction.squeeze(), y.squeeze()) #perda
    batch_loss.backward() 
    optimizer.step() #muda os pesos
    return batch_loss.item()
```

```python
from torch.utils.data import TensorDataset, Dataset, DataLoader

trn_dl = DataLoader(TensorDataset(X_train, y_train)) #dataloader com os dados pra treinar
```

```python
import torch

for epoch in range(2000): #200 epochs
    for ix, batch in enumerate(trn_dl):
        x, y = batch
        x = x.to(device)
        y = y.to(device)
        batch_loss = train_batch(x, y, model, optimizer, criterion) #treina e pega o erro
```
