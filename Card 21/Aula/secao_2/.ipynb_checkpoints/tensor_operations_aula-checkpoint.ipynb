```python
a = torch.ones(2,1,10) #2 pilhas com cada tendo uma linha e 10 colunas
```

```python
a #pega o formato
```

```python
print(f"This is the shape before squeezing: {a.shape}")
squeezed = a.squeeze(1)
print(f"This is the shape after squeezing: {squeezed.shape}") #remove a dimensao de indice 1 se ela tiver tamanho 1
```

```python
a = torch.ones(2,2,10)
print(f"This is the shape before squeezing: {a.shape}")
squeezed = a.squeeze(1)
print(f"This is the shape after squeezing: {squeezed.shape}") #nao consegue remover poois nao tem tamanho 1
```

```python
a = torch.ones(2,3,1)
print(f"This is the shape before squeezing: {a.shape}")
squeezed = a.squeeze(2)
print(f"This is the shape after squeezing: {squeezed.shape}")  #remove pq o 2 indice tem tamanho 1
```

```python
a = torch.ones(2,3)
print(f"This is the shape before unsqueezing: {a.shape}")
squeezed = a.unsqueeze(2)
print(f"This is the shape after unsqueezing: {squeezed.shape}")  #adiciona uma dimensao de tamanho 1
```

```python
a = torch.ones(2,3) #matriz 2x3
```

```python
a #print
```

```python
print(a[None].shape) #cria uma diemnsao nova no indice 0
print(a[:,None].shape) #cria no indice 2
print(a[:,:,None].shape) #e no indice 3
```
