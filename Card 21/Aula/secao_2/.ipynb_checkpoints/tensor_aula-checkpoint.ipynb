```python
a = torch.tensor([[[1,2], [3,4], [5,6], [7,8]]]) #matriz 3D

x_vals = []
y_vals = []
z_vals = [] #listas pra cada valor
for i in range(a.shape[0]):
    for j in range(a.shape[1]):
        for k in range(a.shape[2]): #vai nas dimensoes salvando nas listas
            x_vals.append(i)
            y_vals.append(j)
            z_vals.append(k)

trace = go.Scatter3d( 
    x = x_vals,
    y = y_vals,
    z = z_vals, #pega as listas preenchidas
    mode = 'markers', #so os pontos
    marker = dict(
        size = 5, #tamanho dos pontos
        color = a.flatten(), 
        colorscale = 'Viridis', #cor de cada ponto
        opacity = 0.8 #opacidade
    )
)
    
layout = go.Layout( #define o layout da figura
    margin = dict(l = 0, r = 0, b = 0, t = 0), #margem
    scene = dict( #titulos
        xaxis = dict(title = 'Dimension 1'),
        yaxis = dict(title = 'Dimension 2'),
        zaxis = dict(title = 'Dimension 3')
    )
)

fig = go.Figure(data = [trace], layout = layout) #cria a figura
pyo.plot(fig, filename = 'tensor.html') #e coloca num html
```
