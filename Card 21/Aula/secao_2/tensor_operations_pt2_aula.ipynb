```python
import torch
```

```python
x = torch.tensor([[1, 2, 3, 4], [5, 6, 7, 8]]) #cria a matriz 2x4
y = torch.tensor([[1, 2, 3], #e aq uma 4,3
                  [2, 3, 4],
                  [4, 5, 6],
                  [7, 8, 9]])
```

```python
torch.matmul(x, y) #multiplica elas
```

```python
x = torch.randn(1, 4, 5) #matriz 4x5
z = torch.cat([x, x], axis = 1) #concatena 2 matriz x
print('Concatenated axis 1: ', x.shape, z.shape) #printa o formato de cada um
```

```python
x = torch.randn(3, 20, 10) #cria 3 matrizes 20 linhas e 10 colunas
z = x.permute(2, 0, 1) #muda as dimensoes da matriz
print('Permute dimensions: ', x.shape, z.shape) #printa
```
