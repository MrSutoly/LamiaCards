```python
import keras
import tensorflow as tf #imports
```

```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import utils as np_utils #imports
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data() #carrega os dados
```

```python
X_train = X_train.reshape(X_train.shape[0], 28, 28, 1)
X_test = X_test.reshape(X_test.shape[0], 28, 28, 1) #coloca mais uma camada pra ficar 3D
X_train = X_train.astype('float32')
X_test = X_test.astype('float32') #formata os valores pra float32
X_train /= 255 
X_test /= 255 #coloca os dados no intervalo entre 0 e 1
y_train = np_utils.to_categorical(y_train, 10) 
y_test = np_utils.to_categorical(y_test, 10) #faz one hot encode nas categoria de y
```

```python
classificador = Sequential() #rede sequencial
classificador.add(InputLayer(shape = (28, 28, 1))) #camada de entrada
classificador.add(Conv2D(32, (3, 3), activation = 'relu')) #camada de convolucao
classificador.add(MaxPooling2D(pool_size = (2, 2))) #reduz a imagem pra proxima camada
classificador.add(Flatten()) #joga pra vetor
classificador.add(Dense(units = 128, activation = 'relu')) #camada densa com a funcao relu
classificador.add(Dense(units = 10, activation = 'softmax')) #funcao de saida com o softmax 
classificador.compile(loss = 'categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy']) #compila a rede
```

```python
#gera novas imagens de treino
gerador_treinamento = ImageDataGenerator(rotation_range = 7, horizontal_flip = True, shear_range = 0.2, height_shift_range = 0.07, zoom_range = 0.2)
```

```python
gerar_teste = ImageDataGenerator() #imagens de teste
```

```python
base_treinamento = gerador_treinamento.flow(X_train, y_train, batch_size = 128) #base de treino 
```

```python
base_teste = gerador_treinamento.flow(X_test, y_test, batch_size = 128) #base de teste
```

```python
classificador.fit(base_treinamento, epochs = 5, validation_data = base_teste) #treina o modelo
```
