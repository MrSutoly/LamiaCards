```python
import keras
import numpy as np
import tensorflow as tf
import sklearn #imports
```

```python
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from keras import utils as np_utils
from sklearn.model_selection import StratifiedKFold #imports
```

```python
(X, y), (X_test, y_test) = mnist.load_data() #carrega os dados
X = X.reshape(X.shape[0], 28, 28, 1) #coloca o canal de cor pra ficar 3d
X = X.astype('float32')  #formata com o float32
X /= 255 #normaliza pro intervalo 0 e 1
y = np_utils.to_categorical(y, 10)  #one hot code nas classes de y
```

```python
seed = 5 
np.random.seed(seed) #seeds pra sempre gerar os mesmos numeros
```

```python
kfold = StratifiedKFold(n_splits = 5, shuffle = True, random_state = seed) #definindo o kfold
```

```python
resultados = [] #lista dos resultados
for indice_treinamento, indice_teste in kfold.split(X, np.zeros(shape = (y.shape[0], 1))): #divide os dados de treino e teste
    print("Indices treinamento: ", indice_treinamento, "Indice test: ", indice_teste) #printa os indices de treino e teste
    rede_neural = Sequential() #define como sequencial
    rede_neural.add(InputLayer(shape = (28,28,1))) #camada de entrada
    rede_neural.add(Conv2D(32, (3,3), activation = 'relu')) #camada de convolucao 
    rede_neural.add(MaxPooling2D(pool_size = (2,2))) #reduz a imagem
    rede_neural.add(Flatten()) #passa pra vetor
    rede_neural.add(Dense(units = 128, activation = 'relu')) #camada densa com relu
    rede_neural.add(Dense(units = 10, activation = 'softmax')) #camada de saida com softmax
    rede_neural.compile(loss = 'categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy']) #compila a rede
    rede_neural.fit(X[indice_treinamento], y[indice_treinamento], batch_size = 128, epochs = 5) #treina a rede

    precisao = rede_neural.evaluate(X[indice_teste], y[indice_teste]) #avalia a rede com os dados de teste
    resultados.append(precisao[1]) #coloca na lista a acuracia
```
