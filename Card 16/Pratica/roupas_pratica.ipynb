```python
import numpy as np
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.utils import to_categorical #imports
```

```python
(X_treino, y_treino), (X_teste, y_teste) = fashion_mnist.load_data() #carrega os dados

X_treino = X_treino.reshape(X_treino.shape[0], 28, 28, 1).astype('float32') / 255.0
X_teste = X_teste.reshape(X_teste.shape[0], 28, 28, 1).astype('float32') / 255.0 #normalizando pra 0 e 1
```

```python
y_treino = to_categorical(y_treino, 10)
y_teste = to_categorical(y_teste, 10) #jogando pra dado categorico fazendo o one hot code
```

```python
modelo_roupas = Sequential() #define a ordem sequencial do modelo

modelo_roupas.add(Conv2D(filters = 64, kernel_size = (3, 3), activation = 'relu', input_shape = (28, 28, 1))) #camada de convolucao pra imagem
modelo_roupas.add(MaxPooling2D(pool_size = (2, 2))) #faz o maxpooling pra pegar as caracteristicas
modelo_roupas.add(Dropout(rate = 0.2)) #desativa 20% dos neuronios

modelo_roupas.add(Conv2D(filters = 64, kernel_size = (3, 3), activation = 'relu')) #mais uma camada de convolucao
modelo_roupas.add(MaxPooling2D(pool_size = (2, 2))) #maxpooling novamente
modelo_roupas.add(Flatten()) #colocando os dados pra vetores pra passar pra rede neural

modelo_roupas.add(Dense(units = 128, activation = 'relu')) #camada densa da rede com 128 neuronios
modelo_roupas.add(Dropout(rate = 0.3)) #desativando 30% dos neuronios
modelo_roupas.add(Dense(units = 10, activation = 'softmax')) #camada de saida com a softmax pra classificação

modelo_roupas.compile(loss = 'categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy']) #compila a rede
```

```python
modelo_roupas.summary() #sumario da rede
```

```python
rede = modelo_roupas.fit(X_treino, y_treino, epochs = 5, batch_size = 64, validation_split = 0.2) #treinando
```

```python
resultados = modelo_roupas.evaluate(X_teste, y_teste) #testa o modelo com novas fotos de roupa
```

```python
resultados #loss e acuracia
```

```python
rede  =  modelo_roupas.fit(X_treino, y_treino, epochs  =  10, batch_size  =  128, validation_split  =  0.2) #treina a rede
```

```python
resultado_final = modelo_roupas.evaluate(X_teste, y_teste) #testando a rede
```
